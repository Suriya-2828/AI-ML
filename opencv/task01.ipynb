```python
# Task 1: Color Space Conversion

## Aim

#To convert a low-light image into different color spaces (Grayscale, HSV, and RGB) using OpenCV and compare the outputs.


#Images are generally loaded in BGR format by OpenCV. Different color spaces are useful for different computer vision tasks.

#- **Grayscale** removes color information and keeps only intensity values.
#- **RGB** is the standard color format used for displaying images.
#- **HSV** separates the image into Hue, Saturation, and Value components, making it useful for color analysis and lighting-related applications.

import cv2
import matplotlib.pyplot as plt

# Read the low-light image
img = cv2.imread("/Users/suriya/Documents/opencv/SuriyaKumar/raw_captures/lowlight.jpg")

# Check whether image is loaded correctly
if img is None:
    print("Image not found. Check the file path.")
else:
    print("Image loaded successfully!")

# Convert BGR image to different color spaces

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Split HSV channels

h, s, v = cv2.split(hsv)

plt.figure(figsize=(16,10))

# Original (RGB)
plt.subplot(2,3,1)
plt.imshow(rgb)
plt.title("Original (RGB)")
plt.axis("off")

# Grayscale
plt.subplot(2,3,2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

# HSV Image
plt.subplot(2,3,3)
plt.imshow(hsv)
plt.title("HSV")
plt.axis("off")

# Hue Channel
plt.subplot(2,3,4)
plt.imshow(h, cmap="gray")
plt.title("Hue Channel")
plt.axis("off")

# Saturation Channel
plt.subplot(2,3,5)
plt.imshow(s, cmap="gray")
plt.title("Saturation Channel")
plt.axis("off")

# Value Channel
plt.subplot(2,3,6)
plt.imshow(v, cmap="gray")
plt.title("Value Channel")
plt.axis("off")

plt.tight_layout()
plt.show()

## Observation

#The low-light image was successfully converted into Grayscale, RGB, and HSV color spaces.

#Among the HSV channels, the **Value (V)** channel preserved the details in the darker regions the best. This is because the Value channel represents the brightness of each pixel. Even though the image was captured in low light, the brightness information still retained the visible features of the subject.

#The Hue channel contained only color information and lost many dark details, while the Saturation channel showed the intensity of colors but was less useful for identifying objects in poorly illuminated regions.

## Conclusion

#In this task, the low-light image was converted into different color spaces using OpenCV. The comparison showed that different color spaces highlight different image characteristics. The HSV Value channel was the most useful for observing details in the darker parts of the image.
```
