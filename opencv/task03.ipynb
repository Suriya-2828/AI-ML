```python
# Task 3: Geometric Transformations

## Aim

#To apply translation, rotation, scaling, and affine transformation on an image using OpenCV and observe how each transformation changes the image.

## Theory

#Geometric transformations change the position, orientation, or size of an image.

#- Translation shifts the image horizontally and vertically.
#- Rotation turns the image around a fixed point.
#- Scaling enlarges or shrinks the image.
#- Affine transformation changes the image while preserving parallel lines.

## Reference Number

#Last two digits of phone number: 75

#Rotation Angle = 75 × 3 = 225°

#Scale Factor = 1 + (5 / 10) = 1.5

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("/Users/suriya/Documents/opencv/SuriyaKumar/raw_captures/objects.jpg")

if img is None:
    print("Image not found.")
else:
    print("Image loaded successfully.")

rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

height, width = img.shape[:2]

translation_matrix = np.float32([
    [1, 0, 100],
    [0, 1, 50]
])

translated = cv2.warpAffine(
    img,
    translation_matrix,
    (width, height)
)

angle = 225
scale = 1.5

rotation_matrix = cv2.getRotationMatrix2D(
    (width//2, height//2),
    angle,
    scale
)

rotated = cv2.warpAffine(
    img,
    rotation_matrix,
    (width, height)
)

scaled = cv2.resize(
    img,
    None,
    fx=scale,
    fy=scale
)

pts1 = np.float32([
    [50,50],
    [200,50],
    [50,200]
])

pts2 = np.float32([
    [30,70],
    [220,50],
    [80,250]
])

affine_matrix = cv2.getAffineTransform(
    pts1,
    pts2
)

affine = cv2.warpAffine(
    img,
    affine_matrix,
    (width, height)
)

images = [
    rgb,
    cv2.cvtColor(translated, cv2.COLOR_BGR2RGB),
    cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB),
    cv2.cvtColor(affine, cv2.COLOR_BGR2RGB)
]

titles = [
    "Original",
    "Translated",
    "Rotated (225°)",
    "Affine Transform"
]

plt.figure(figsize=(15,10))

for i in range(4):
    plt.subplot(2,2,i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()

plt.figure(figsize=(7,7))
plt.imshow(cv2.cvtColor(scaled, cv2.COLOR_BGR2RGB))
plt.title("Scaled Image (1.5x)")
plt.axis("off")
plt.show()

print("Reference Number :", 75)
print("Rotation Angle :", angle)
print("Scale Factor :", scale)

## Observation

#The objects image was successfully transformed using different geometric operations.

#- Translation shifted the image 100 pixels to the right and 50 pixels downward.
#- Rotation turned the image by 225° around its center.
#- Scaling enlarged the image by a factor of 1.5.
#- The affine transformation changed the shape of the image while preserving parallel lines.

#Each transformation produced a different visual effect without changing the original image data.

## Conclusion

#In this task, different geometric transformations were applied using OpenCV. Translation changed the image position, rotation changed its orientation, scaling changed its size, and affine transformation modified its geometry. These transformations are commonly used in image processing and computer vision applications.

```
