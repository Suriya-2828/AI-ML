```python
# Task 9: Real-Time Video Processing

## Aim

#To process a video frame by frame using OpenCV, apply a grayscale effect, and save the processed output using VideoWriter.

## Theory

#A video consists of a sequence of image frames. OpenCV reads each frame individually using VideoCapture. Each frame can be processed independently before being displayed or written to a new video using VideoWriter.

import cv2
cap = cv2.VideoCapture("/Users/suriya/Documents/opencv/SuriyaKumar/raw_captures/video.mp4")

if not cap.isOpened():
    print("Error opening video.")
else:
    print("Video opened successfully.")

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

print("Width :", width)
print("Height:", height)
print("FPS   :", fps)

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

out = cv2.VideoWriter(
    "output/grayscale_video.mp4",
    fourcc,
    fps,
    (width, height),
    False
)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    cv2.imshow("Grayscale Video", gray)

    out.write(gray)

    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

cap.release()
out.release()

cv2.destroyAllWindows()

print("Video processing completed.")

## Observation

#The input video was successfully processed frame by frame. Each frame was converted to grayscale and displayed in real time. The processed video was also saved successfully using OpenCV's VideoWriter.

## Bug Encountered

#Initially, the output video was not saved correctly because the frame size passed to VideoWriter did not match the input video resolution.

#After checking the width and height using VideoCapture properties and using the same values in VideoWriter, the processed video was saved correctly.

## Conclusion

#In this task, a video was processed frame by frame using OpenCV. A grayscale effect was applied to each frame, and the processed frames were saved into a new video file. This demonstrates the basic workflow for real-time video processing.
```
