```python
# Task 8: Face Detection

## Aim

#To detect human faces in an image using OpenCV's Haar Cascade Classifier.

## Theory

#A Haar Cascade Classifier is a machine learning-based object detection method used to detect faces in images. It scans different regions of the image and identifies areas that resemble human facial features.

import cv2
import matplotlib.pyplot as plt
img = cv2.imread("/Users/suriya/Documents/opencv/SuriyaKumar/raw_captures/normal.jpg")

if img is None:
    print("Image not found.")
else:
    print("Image loaded successfully.")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades +
    "haarcascade_frontalface_default.xml"
)
faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.1,
    minNeighbors=5,
    minSize=(30,30)
)

print("Number of Faces Detected:", len(faces))

output = img.copy()

for (x, y, w, h) in faces:
    cv2.rectangle(
        output,
        (x, y),
        (x+w, y+h),
        (0,255,0),
        2
    )

plt.figure(figsize=(8,8))

plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
plt.title("Detected Face")
plt.axis("off")

plt.show()

for i, (x, y, w, h) in enumerate(faces):
    print(f"Face {i+1}")
    print(f"X = {x}")
    print(f"Y = {y}")
    print(f"Width = {w}")
    print(f"Height = {h}")

## Observation

#The Haar Cascade classifier successfully detected the face in the normal-light image.

#One face was identified, and a bounding rectangle was drawn around it. The detection worked well because the face was clearly visible and the lighting conditions were sufficient for the classifier to identify facial features.
```
