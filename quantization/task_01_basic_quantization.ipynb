```python
# Task 1: Tensor Quantization and Dequantization

## Objective

#The objective of this task is to understand the basic process of affine INT8 quantization and dequantization using NumPy.

#In this task, the scale and zero point are already provided. We use these values to:
#- Convert a floating-point tensor into an INT8 tensor (Quantization).
#- Convert the INT8 tensor back into floating-point values (Dequantization).
#- Measure the error introduced due to quantization.

#Finally, we calculate different error metrics such as:
#- Absolute Error
#- Mean Absolute Error (MAE)
#- Maximum Absolute Error

import numpy as np
# Input tensor
tensor = np.array([1.0, -0.5, 3.2, -2.8, 0.0], dtype=np.float32)

# Given quantization parameters
scale = 0.025
zero_point = 0

def quantize_tensor(tensor, scale, zero_point):
    """
    Quantizes a floating-point tensor into INT8 values.
    """

    # Divide by scale
    scaled_tensor = tensor / scale

    # Round to nearest integer
    rounded_tensor = np.round(scaled_tensor)

    # Add zero point
    shifted_tensor = rounded_tensor + zero_point

    # Clip to INT8 range
    clipped_tensor = np.clip(shifted_tensor, -128, 127)

    # Convert to INT8
    return clipped_tensor.astype(np.int8)

def dequantize_tensor(quantized_tensor, scale, zero_point):
    """
    Converts INT8 tensor back to floating-point values.
    """

    return (quantized_tensor.astype(np.float32) - zero_point) * scale

# Quantize the tensor
quantized_tensor = quantize_tensor(tensor, scale, zero_point)

# Dequantize the tensor
dequantized_tensor = dequantize_tensor(quantized_tensor, scale, zero_point)

# Calculate error
absolute_error = np.abs(tensor - dequantized_tensor)

mae = np.mean(absolute_error)
max_error = np.max(absolute_error)

print("Original Tensor:")
print(tensor)

print("\nQuantized Tensor (INT8):")
print(quantized_tensor)

print("\nDequantized Tensor:")
print(dequantized_tensor)

print("\nAbsolute Error:")
print(absolute_error)

print("\nMean Absolute Error (MAE):")
print(mae)

print("\nMaximum Absolute Error:")
print(max_error)

## Observations

#- The floating-point tensor was successfully converted into INT8 values.
#- During dequantization, most values were reconstructed exactly.
#- The value **3.2** exceeded the maximum representable value for the given scale and was clipped to **127**.
#- Due to clipping, the reconstructed value became **3.175**, introducing a small quantization error.
#- The Mean Absolute Error (MAE) is very small, indicating that the quantization process preserved the original values well.
#- The maximum error occurred only for the clipped value.

## Conclusion

#In this task, affine INT8 quantization and dequantization were implemented using NumPy. The original tensor was converted into INT8 format and reconstructed back into floating-point values. The quantization process introduced only a small error due to clipping of a value outside the representable range. This demonstrates that INT8 quantization can significantly reduce precision while still maintaining values close to the original tensor.
```
